```python
top_countries = df['City'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Cits with Most Restaurants')
plt.xlabel('Number of Restaurants')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Rating'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

```python
top_cuisines = df['Cuisines'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(y=top_cuisines.index, x=top_cuisines.values, palette='magma')
plt.title('Top 10 Most Common Cuisines')
plt.xlabel('Frequency')
plt.ylabel('Cuisines')
plt.show()
```

```python
online_delivery = df['Has Online delivery'].value_counts()
plt.pie(online_delivery, labels=online_delivery.index, autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#ff9999'])
plt.title('Online Delivery Availability')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['Votes', 'Average Cost for two', 'Rating']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
ratings = final_df.groupby(['Rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings.head()
```

Observations:
1. When rating is between 4.5 - 4.9 --> Excellent
2. When rating is between 4.0 - 3.4 --> Very Good
3. When rating is between 3.5 - 3.9 --> Good
4. When rating is between 3.0 - 3.4 --> Average
5. When rating is between 2.5 - 2.9 --> Average
6. When rating is between 2.0 - 2.4 --> Poor

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(12,6)
sns.barplot(x="Rating",y="Rating Count",data=ratings)
```

```python
sns.barplot(x="Rating",y="Rating Count",hue='Rating color',data=ratings,palette=['blue','red','orange','yellow','green','green'])
```

Observation:
1. Not Rated Count is very high.
2. Maximum number of rating are between 2.5 to 3.4
